```python
# ─────────────────────────────────────────────────────────────────────────────
# CHARGEMENT
# ─────────────────────────────────────────────────────────────────────────────
DATA_PATH = 'data/'  # ← Adapter si nécessaire

matches     = pd.read_csv(DATA_PATH + 'matchs_2013_2024.csv')
m2025       = pd.read_csv(DATA_PATH + 'match_2025.csv')
clubs       = pd.read_csv(DATA_PATH + 'clubs_fr.csv')
valuation   = pd.read_csv(DATA_PATH + 'player_valuation_before_season.csv')
events      = pd.read_csv(DATA_PATH + 'game_events_before2025.csv')
lineups     = pd.read_csv(DATA_PATH + 'game_lineups.csv', low_memory=False)
appearances = pd.read_csv(DATA_PATH + 'player_appearance.csv')
sample      = pd.read_csv(DATA_PATH + 'sample_results.csv')

for df in [matches, m2025, appearances]:
    df['date'] = pd.to_datetime(df['date'])
valuation['date'] = pd.to_datetime(valuation['date'])
events['date']    = pd.to_datetime(events['date'])

# Nettoyage transferts
def parse_transfer(val):
    if pd.isna(val): return 0.0
    val = str(val).replace('€','').replace('+','').strip()
    if val.endswith('m'): return float(val[:-1]) * 1e6
    if val.endswith('k'): return float(val[:-1]) * 1e3
    try: return float(val)
    except: return 0.0

clubs['net_transfer_eur'] = clubs['net_transfer_record'].apply(parse_transfer)

# Valuation par club/saison
valuation['season'] = valuation['date'].dt.year - 1
club_val_season = valuation.groupby(['current_club_id','season'])['market_value_in_eur'].agg(
    squad_total_value='sum', squad_avg_value='mean'
).reset_index().rename(columns={'current_club_id':'club_id'})

def add_valuation_features(df, val_df):
    df = df.merge(val_df.rename(columns={
        'club_id':'home_club_id','squad_total_value':'home_squad_value','squad_avg_value':'home_avg_player_value'
    }), on=['home_club_id','season'], how='left')
    df = df.merge(val_df.rename(columns={
        'club_id':'away_club_id','squad_total_value':'away_squad_value','squad_avg_value':'away_avg_player_value'
    }), on=['away_club_id','season'], how='left')
    df['squad_value_ratio'] = df['home_squad_value'] / (df['away_squad_value'] + 1)
    df['squad_value_diff']  = df['home_squad_value'] - df['away_squad_value']
    return df

matches = add_valuation_features(matches, club_val_season)
print('✅ Données chargées et prétraitées')
```

## AMÉLIORATION 1 — Proxies xG depuis `game_events`

### Contexte

Les données **Opta/Statsbomb** (xG, possession, tirs) ne sont pas disponibles directement. Cependant, notre fichier `game_events` contient des **informations équivalentes** encodées dans la colonne `description` des buts :

- Type de tir : pied droit, pied gauche, tête, direct coup-franc
- Buts sur penalty (difficulté très faible → xG ≈ 0.79)
- Buts en contre-attaque (difficiles à défendre)
- Buts de longue distance (difficulté élevée → xG faible)
- Timing : buts précoces vs tardifs (momentum)
- Cartons : fréquence et type (proxy de l'agressivité/discipline)

Ces features **approximent le xG** sans accès aux données propriétaires.

```python
# ─────────────────────────────────────────────────────────────────────────────
# EXTRACTION DES PROXIES xG DEPUIS game_events
# ─────────────────────────────────────────────────────────────────────────────

goals_ev = events[events['type'] == 'Goals'].copy()
cards_ev  = events[events['type'] == 'Cards'].copy()
subs_ev   = events[events['type'] == 'Substitutions'].copy()

# ── Parsing des types de buts ──────────────────────────────────────────────
goals_ev['shot_type']       = goals_ev['description'].str.extract(r',\s*([^,]+),')[0].str.strip()
goals_ev['is_penalty']      = goals_ev['description'].str.contains('Penalty', case=False, na=False).astype(int)
goals_ev['is_header']       = goals_ev['description'].str.contains('Header', case=False, na=False).astype(int)
goals_ev['is_freekick']     = goals_ev['description'].str.contains('Direct free kick', case=False, na=False).astype(int)
goals_ev['is_counter']      = goals_ev['description'].str.contains('Counter attack', case=False, na=False).astype(int)
goals_ev['is_longdist']     = goals_ev['description'].str.contains('Long distance', case=False, na=False).astype(int)
goals_ev['is_own_goal']     = goals_ev['description'].str.contains('Own-goal', case=False, na=False).astype(int)
goals_ev['is_early']        = (goals_ev['minute'] <= 15).astype(int)
goals_ev['is_late']         = (goals_ev['minute'] >= 75).astype(int)

# xG approximatif par type (valeurs issues de la littérature)
XG_MAP = {
    'Penalty'            : 0.79,
    'Tap-in'             : 0.60,
    'Right-footed shot'  : 0.10,
    'Left-footed shot'   : 0.09,
    'Header'             : 0.08,
    'Direct free kick'   : 0.06,
    'Deflected shot on goal': 0.08,
    'Counter attack goal': 0.12,
    'Long distance kick' : 0.04,
    'Solo run'           : 0.15,
}
goals_ev['approx_xg'] = goals_ev['shot_type'].map(XG_MAP).fillna(0.09)

print('Distribution des types de buts :')
print(goals_ev['shot_type'].value_counts().head(10))
print(f'\nxG moyen approximatif : {goals_ev["approx_xg"].mean():.3f}')
```

```python
# ── Agrégation des proxies xG par match et par club ────────────────────────

def agg_xg_per_match_club(goals_df):
    """Agrège les métriques de qualité de buts par (game_id, club_id)."""
    return goals_df.groupby(['game_id', 'club_id']).agg(
        nb_goals          = ('is_penalty', 'count'),
        nb_penalties      = ('is_penalty', 'sum'),
        nb_headers        = ('is_header', 'sum'),
        nb_freekicks      = ('is_freekick', 'sum'),
        nb_counter        = ('is_counter', 'sum'),
        nb_longdist       = ('is_longdist', 'sum'),
        nb_own_goals      = ('is_own_goal', 'sum'),
        nb_early_goals    = ('is_early', 'sum'),
        nb_late_goals     = ('is_late', 'sum'),
        total_approx_xg   = ('approx_xg', 'sum'),
        avg_goal_minute   = ('minute', 'mean'),
    ).reset_index()

xg_per_match = agg_xg_per_match_club(goals_ev)

# Merge sur matches (domicile et extérieur)
xg_home = xg_per_match.rename(columns={
    'club_id': 'home_club_id',
    **{c: f'home_{c}' for c in xg_per_match.columns if c not in ['game_id','club_id']}
})
xg_away = xg_per_match.rename(columns={
    'club_id': 'away_club_id',
    **{c: f'away_{c}' for c in xg_per_match.columns if c not in ['game_id','club_id']}
})

matches = matches.merge(xg_home, on=['game_id','home_club_id'], how='left')
matches = matches.merge(xg_away, on=['game_id','away_club_id'], how='left')

# Proxy xG différentiel
matches['xg_diff']         = matches['home_total_approx_xg'].fillna(0) - matches['away_total_approx_xg'].fillna(0)
matches['penalty_diff']    = matches['home_nb_penalties'].fillna(0)    - matches['away_nb_penalties'].fillna(0)
matches['late_goal_diff']  = matches['home_nb_late_goals'].fillna(0)   - matches['away_nb_late_goals'].fillna(0)

print('✅ Features proxies xG ajoutées au dataset')
print(f'Shape matches : {matches.shape}')
```

```python
# ── Cartons par match et par club ──────────────────────────────────────────

cards_ev['is_red']    = cards_ev['description'].str.contains('Red|red', na=False).astype(int)
cards_ev['is_yellow'] = (cards_ev['is_red'] == 0).astype(int)
cards_ev['foul_type'] = cards_ev['description'].str.extract(r',\s*([^,]+)')[0].str.strip()

cards_agg = cards_ev.groupby(['game_id','club_id']).agg(
    nb_cards_total = ('game_event_id','count'),
    nb_red_cards   = ('is_red','sum'),
    nb_yellow_cards= ('is_yellow','sum'),
).reset_index()

cards_home = cards_agg.rename(columns={
    'club_id':'home_club_id',
    **{c:f'home_{c}' for c in cards_agg.columns if c not in ['game_id','club_id']}
})
cards_away = cards_agg.rename(columns={
    'club_id':'away_club_id',
    **{c:f'away_{c}' for c in cards_agg.columns if c not in ['game_id','club_id']}
})

matches = matches.merge(cards_home, on=['game_id','home_club_id'], how='left')
matches = matches.merge(cards_away, on=['game_id','away_club_id'], how='left')
matches['discipline_diff'] = (matches['home_nb_cards_total'].fillna(0) -
                               matches['away_nb_cards_total'].fillna(0))

print('✅ Features cartons et discipline ajoutées')
```

```python
# ── Visualisation des proxies xG ──────────────────────────────────────────

fig, axes = plt.subplots(1, 3, figsize=(16, 5))

colors_res = {1:'#2ecc71', 0:'#f39c12', -1:'#e74c3c'}
labels_res = {1:'Dom. gagne', 0:'Nul', -1:'Ext. gagne'}

for ax, feat, title in zip(axes,
    ['xg_diff', 'late_goal_diff', 'discipline_diff'],
    ['Différentiel xG approx.', 'Différentiel buts tardifs (>75min)', 'Différentiel cartons']
):
    for res in [1, 0, -1]:
        sub = matches[matches['results'] == res][feat].dropna()
        sub.plot(kind='kde', ax=ax, label=labels_res[res],
                 color=colors_res[res], linewidth=2)
    ax.set_title(title, fontweight='bold')
    ax.set_xlabel(feat)
    ax.legend(fontsize=9)

plt.suptitle('Proxies xG — Pouvoir discriminant par résultat',
             fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

**💡 Insight :** Le **différentiel xG approximatif** (buts marqués pondérés par leur difficulté) montre une séparation claire entre les 3 classes : les équipes gagnantes produisent des opportunités de meilleure qualité. Les buts tardifs et la discipline apportent un signal complémentaire sur la dynamique du match.

> ⚠️ Ces features sont calculées **après le match** — elles seront utilisées en features **glissantes** (moyenne des 5 derniers matchs) pour la prédiction.

```python
# ── Features glissantes xG sur les 5 derniers matchs ─────────────────────

def compute_rolling_xg(matches_df, n=5):
    """Pour chaque match, calcule les moyennes xG des n matchs précédents."""
    df = matches_df.sort_values('date').copy()
    rows = []
    
    xg_cols = {
        'home': ['home_total_approx_xg','home_nb_penalties','home_nb_headers',
                 'home_nb_late_goals','home_nb_cards_total'],
        'away': ['away_total_approx_xg','away_nb_penalties','away_nb_headers',
                 'away_nb_late_goals','away_nb_cards_total']
    }
    
    for _, row in df.iterrows():
        hid = row['home_club_id']; aid = row['away_club_id']; date = row['date']
        
        def get_prev(cid):
            mask = ((df['home_club_id']==cid)|(df['away_club_id']==cid)) & (df['date']<date)
            prev = df[mask].tail(n)
            if len(prev) == 0:
                return {}
            res = {}
            for col_h, col_a in zip(xg_cols['home'], xg_cols['away']):
                vals = []
                for _, m in prev.iterrows():
                    if m['home_club_id'] == cid:
                        vals.append(m.get(col_h, 0) or 0)
                    else:
                        vals.append(m.get(col_a, 0) or 0)
                res[col_h.replace('home_','rolling_')] = np.mean(vals) if vals else 0
            return res
        
        rh = get_prev(hid)
        ra = get_prev(aid)
        entry = {'game_id': row['game_id']}
        for k, v in rh.items(): entry[f'home_{k}'] = v
        for k, v in ra.items(): entry[f'away_{k}'] = v
        rows.append(entry)
    
    return pd.DataFrame(rows)

print('⏳ Calcul des rolling xG features...')
rolling_xg_df = compute_rolling_xg(matches)
print(f'✅ Rolling xG calculé : {rolling_xg_df.shape}')
display(rolling_xg_df.head(3))
```

## AMÉLIORATION 2 — LSTM (Modèle de Séries Temporelles)

### Contexte

Un **LSTM (Long Short-Term Memory)** est un réseau de neurones récurrent conçu pour capturer les **dépendances temporelles à long terme**. Contrairement aux modèles classiques qui voient chaque match indépendamment, le LSTM :

- Reçoit en entrée une **séquence des N derniers matchs** d'une équipe
- Apprend à détecter des **patterns temporels** (séries de victoires, crises de résultats, forme montante)
- Peut capturer des **effets de mémoire** (un mauvais résultat il y a 3 matchs impacte-t-il encore aujourd'hui ?)

**Architecture :**
- Input : séquence de 10 matchs × features par équipe
- 2 couches LSTM avec Dropout
- Dense + Softmax (3 classes)

```python
# ─────────────────────────────────────────────────────────────────────────────
# PRÉPARATION DES SÉQUENCES POUR LE LSTM
# ─────────────────────────────────────────────────────────────────────────────

SEQ_LEN = 10  # Nombre de matchs en historique

# Features par match (côté équipe)
LSTM_FEATURES = [
    'goals_scored', 'goals_conceded',
    'won', 'drawn', 'lost',
    'is_home',
    'opp_position',
]
N_FEAT = len(LSTM_FEATURES)

def build_lstm_sequences(matches_df, seq_len=SEQ_LEN):
    """
    Construit les séquences (X_home, X_away) et les labels y pour le LSTM.
    Pour chaque match, on extrait les seq_len matchs précédents de chaque équipe.
    """
    df = matches_df.sort_values('date').copy().reset_index(drop=True)
    
    def get_team_sequence(club_id, before_idx):
        """Séquence des seq_len matchs d'un club avant l'index donné."""
        prev = df[
            (df.index < before_idx) &
            ((df['home_club_id'] == club_id) | (df['away_club_id'] == club_id))
        ].tail(seq_len)
        
        seq = []
        for _, m in prev.iterrows():
            is_home = int(m['home_club_id'] == club_id)
            gf = m['home_club_goals'] if is_home else m['away_club_goals']
            ga = m['away_club_goals'] if is_home else m['home_club_goals']
            res = m['results']
            won  = int((is_home and res==1) or (not is_home and res==-1))
            draw = int(res==0)
            lost = int((is_home and res==-1) or (not is_home and res==1))
            opp_pos = m['away_club_position'] if is_home else m['home_club_position']
            seq.append([gf, ga, won, draw, lost, is_home, opp_pos / 20.0])
        
        # Padding si moins de seq_len matchs disponibles
        while len(seq) < seq_len:
            seq.insert(0, [0.0]*N_FEAT)  # Pad au début
        
        return np.array(seq[-seq_len:], dtype=np.float32)
    
    X_home, X_away, y_lstm, game_ids = [], [], [], []
    label_map = {-1: 0, 0: 1, 1: 2}
    
    for idx, row in df.iterrows():
        if idx < seq_len:  # Skip les premiers matchs sans historique suffisant
            continue
        seq_h = get_team_sequence(row['home_club_id'], idx)
        seq_a = get_team_sequence(row['away_club_id'], idx)
        X_home.append(seq_h)
        X_away.append(seq_a)
        y_lstm.append(label_map[row['results']])
        game_ids.append(row['game_id'])
    
    return (np.array(X_home), np.array(X_away),
            np.array(y_lstm), np.array(game_ids))

print(f'⏳ Construction des séquences LSTM (seq_len={SEQ_LEN})...')
X_lstm_home, X_lstm_away, y_lstm, game_ids_lstm = build_lstm_sequences(matches)
print(f'✅ Séquences construites :')
print(f'   X_home : {X_lstm_home.shape}  (matchs × seq × features)')
print(f'   X_away : {X_lstm_away.shape}')
print(f'   y      : {y_lstm.shape} | classes : {np.unique(y_lstm, return_counts=True)}')
```

```python
# ─────────────────────────────────────────────────────────────────────────────
# ARCHITECTURE DU MODÈLE LSTM
# ─────────────────────────────────────────────────────────────────────────────

# Split temporel (même logique que les autres modèles)
match_seasons = matches.sort_values('date').iloc[SEQ_LEN:]['season'].values
train_mask_lstm = match_seasons < 2024
test_mask_lstm  = match_seasons >= 2024

Xh_tr = X_lstm_home[train_mask_lstm]; Xh_te = X_lstm_home[test_mask_lstm]
Xa_tr = X_lstm_away[train_mask_lstm]; Xa_te = X_lstm_away[test_mask_lstm]
y_tr_l= y_lstm[train_mask_lstm];      y_te_l= y_lstm[test_mask_lstm]

y_tr_cat = to_categorical(y_tr_l, 3)
y_te_cat = to_categorical(y_te_l, 3)

# ── Architecture Double-LSTM (une branche par équipe) ───────────────────
from tensorflow.keras.layers import Input, Concatenate
from tensorflow.keras.models import Model

def build_lstm_model(seq_len, n_features):
    """Modèle LSTM avec deux branches symétriques (domicile / extérieur)."""
    
    # Branche équipe domicile
    inp_home = Input(shape=(seq_len, n_features), name='home_seq')
    x_h = LSTM(64, return_sequences=True, name='lstm_home_1')(inp_home)
    x_h = Dropout(0.3)(x_h)
    x_h = LSTM(32, name='lstm_home_2')(x_h)
    x_h = BatchNormalization()(x_h)
    
    # Branche équipe extérieure
    inp_away = Input(shape=(seq_len, n_features), name='away_seq')
    x_a = LSTM(64, return_sequences=True, name='lstm_away_1')(inp_away)
    x_a = Dropout(0.3)(x_a)
    x_a = LSTM(32, name='lstm_away_2')(x_a)
    x_a = BatchNormalization()(x_a)
    
    # Fusion
    merged = Concatenate()([x_h, x_a])
    x = Dense(64, activation='relu')(merged)
    x = Dropout(0.3)(x)
    x = Dense(32, activation='relu')(x)
    out = Dense(3, activation='softmax', name='output')(x)
    
    model = Model(inputs=[inp_home, inp_away], outputs=out)
    model.compile(
        optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
        loss='categorical_crossentropy',
        metrics=['accuracy']
    )
    return model

lstm_model = build_lstm_model(SEQ_LEN, N_FEAT)
lstm_model.summary()
```

```python
# ─────────────────────────────────────────────────────────────────────────────
# ENTRAÎNEMENT DU LSTM
# ─────────────────────────────────────────────────────────────────────────────

callbacks = [
    EarlyStopping(monitor='val_accuracy', patience=15, restore_best_weights=True, verbose=1),
    ReduceLROnPlateau(monitor='val_loss', factor=0.5, patience=7, min_lr=1e-5, verbose=1)
]

print('⏳ Entraînement du LSTM...')
history = lstm_model.fit(
    [Xh_tr, Xa_tr], y_tr_cat,
    validation_data=([Xh_te, Xa_te], y_te_cat),
    epochs=100,
    batch_size=64,
    callbacks=callbacks,
    verbose=0
)

# Évaluation
lstm_preds_proba = lstm_model.predict([Xh_te, Xa_te], verbose=0)
lstm_preds       = np.argmax(lstm_preds_proba, axis=1)
label_unmap      = {0: -1, 1: 0, 2: 1}
lstm_preds_orig  = np.array([label_unmap[p] for p in lstm_preds])
y_te_orig        = np.array([label_unmap[p] for p in y_te_l])

acc_lstm = accuracy_score(y_te_orig, lstm_preds_orig)
print(f'\n✅ LSTM Test Accuracy : {acc_lstm:.4f}')
```

```python
# ── Courbes d'apprentissage ───────────────────────────────────────────────

fig, axes = plt.subplots(1, 2, figsize=(13, 4))

axes[0].plot(history.history['accuracy'],    label='Train',      color='#3498db', linewidth=2)
axes[0].plot(history.history['val_accuracy'],label='Validation', color='#e74c3c', linewidth=2)
axes[0].set_title('LSTM — Accuracy', fontweight='bold')
axes[0].set_xlabel('Époque'); axes[0].set_ylabel('Accuracy')
axes[0].legend()

axes[1].plot(history.history['loss'],     label='Train',      color='#3498db', linewidth=2)
axes[1].plot(history.history['val_loss'], label='Validation', color='#e74c3c', linewidth=2)
axes[1].set_title('LSTM — Loss (cross-entropy)', fontweight='bold')
axes[1].set_xlabel('Époque'); axes[1].set_ylabel('Loss')
axes[1].legend()

plt.suptitle('Courbes d\'apprentissage du LSTM Double-Branche',
             fontsize=13, fontweight='bold')
plt.tight_layout()
plt.show()

print(classification_report(y_te_orig, lstm_preds_orig,
                             target_names=['Ext. gagne','Nul','Dom. gagne']))
```

**💡 Analyse LSTM :** Le modèle LSTM capture des patterns temporels inaccessibles aux modèles classiques, notamment les **séries de résultats consécutifs** (momentum) et la **dynamique de la forme récente**. L'architecture double-branche traite symétriquement les deux équipes, reflétant la nature duelle d'un match de football.

## AMÉLIORATION 3 — Absence des Joueurs Clés

### Contexte

L'**absence d'un joueur clé** (attaquant principal, gardien, milieu organisateur) peut significativement réduire le niveau d'une équipe. Sans accès à des données de scraping en temps réel, nous inférons l'absence depuis les `game_lineups` :

1. Identifier les **joueurs clés** de chaque club (top 3 par impact score sur toute la saison)
2. Pour chaque match, vérifier si ces joueurs étaient **présents en titulaire**
3. Calculer un **score d'absence** qui reflète la perte potentielle pour l'équipe

```python
# ─────────────────────────────────────────────────────────────────────────────
# FEATURE 3 : ABSENCE DES JOUEURS CLÉS
# ─────────────────────────────────────────────────────────────────────────────

# ── Étape 1 : Score d'impact par joueur et par club ───────────────────────
player_impact = appearances.groupby(['player_id','player_club_id']).agg(
    total_goals   = ('goals',   'sum'),
    total_assists = ('assists', 'sum'),
    total_apps    = ('game_id', 'count'),
    avg_mins      = ('minutes_played', 'mean')
).reset_index()

# Impact = buts + 0.7*assists (pondération standard)
player_impact['impact_score'] = (
    player_impact['total_goals'] + 0.7 * player_impact['total_assists']
)

# Normaliser par le nombre d'apparitions (impact par match)
player_impact['impact_per_game'] = (
    player_impact['impact_score'] / player_impact['total_apps'].clip(lower=1)
)

# ── Étape 2 : Top 5 joueurs clés par club (par impact total) ─────────────
key_players = (
    player_impact[player_impact['total_apps'] >= 5]  # Min 5 matchs
    .sort_values(['player_club_id','impact_score'], ascending=[True, False])
    .groupby('player_club_id').head(5)
)

print(f'Joueurs clés identifiés : {len(key_players)} (pour {key_players["player_club_id"].nunique()} clubs)')
print('\nTop 10 joueurs les plus impactants :')
top10 = player_impact.nlargest(10,'impact_score')[['player_id','player_club_id','total_goals','total_assists','impact_score','total_apps']]
display(top10)
```

```python
# ── Étape 3 : Calcul du score d'absence pour chaque match ─────────────────

# Titulaires par match
starters = lineups[lineups['type'] == 'starting_lineup'][['game_id','player_id','club_id']]

# Set des titulaires par (game_id, club_id) pour recherche rapide
starters_set = starters.groupby(['game_id','club_id'])['player_id'].apply(set).reset_index()
starters_set.columns = ['game_id','club_id','starters_set']

# Joueurs clés par club (dict pour accès rapide)
key_by_club = key_players.groupby('player_club_id').apply(
    lambda g: list(zip(g['player_id'], g['impact_score']))
).to_dict()

def compute_absence_score(game_id, club_id, starters_lookup, key_lookup):
    """Score d'absence = somme des impact_scores des joueurs clés absents."""
    key = key_lookup.get(club_id, [])
    if not key:
        return 0.0, 0
    
    match_starters = starters_lookup.get((game_id, club_id), set())
    if not match_starters:
        return 0.0, 0  # Pas d'info lineup pour ce match
    
    absent_score = 0.0
    nb_absent    = 0
    for pid, imp in key:
        if pid not in match_starters:
            absent_score += imp
            nb_absent    += 1
    return absent_score, nb_absent

# Dictionnaire rapide (game_id, club_id) → set de player_ids
starters_dict = starters.groupby(['game_id','club_id'])['player_id'].apply(set).to_dict()

# Calcul pour tous les matchs
print('⏳ Calcul des scores d\'absence...')
absence_rows = []
for _, row in matches.iterrows():
    gid = row['game_id']
    hid = row['home_club_id']
    aid = row['away_club_id']
    
    h_abs_score, h_nb_abs = compute_absence_score(gid, hid, starters_dict, key_by_club)
    a_abs_score, a_nb_abs = compute_absence_score(gid, aid, starters_dict, key_by_club)
    
    absence_rows.append({
        'game_id'              : gid,
        'home_absence_score'   : h_abs_score,
        'away_absence_score'   : a_abs_score,
        'home_nb_absent_key'   : h_nb_abs,
        'away_nb_absent_key'   : a_nb_abs,
        'absence_score_diff'   : h_abs_score - a_abs_score,  # Positif = domicile plus affaibli
        'has_lineup_info'      : int(h_abs_score > 0 or a_abs_score > 0)
    })

absence_df = pd.DataFrame(absence_rows)
print(f'✅ Scores d\'absence calculés : {absence_df.shape}')
print(f'   Matchs avec info lineup : {absence_df["has_lineup_info"].sum()}')
display(absence_df[absence_df['has_lineup_info']==1].head(5))
```

```python
# ── Visualisation de l'impact des absences ────────────────────────────────

absence_with_result = absence_df.merge(
    matches[['game_id','results']], on='game_id'
)[absence_df.merge(matches[['game_id','results']], on='game_id')['has_lineup_info']==1]

fig, axes = plt.subplots(1, 2, figsize=(13, 5))

# Score d'absence dom vs résultat
absence_with_result.boxplot(
    column='home_absence_score', by='results', ax=axes[0]
)
axes[0].set_title('Score d\'absence domicile selon le résultat', fontweight='bold')
axes[0].set_xlabel('Résultat (1=Dom, 0=Nul, -1=Ext)')
axes[0].set_ylabel('Impact score des absents')

# Différentiel
for res, label, color in [(-1,'Ext. gagne','#e74c3c'),(0,'Nul','#f39c12'),(1,'Dom. gagne','#2ecc71')]:
    sub = absence_with_result[absence_with_result['results']==res]['absence_score_diff'].dropna()
    sub.plot(kind='kde', ax=axes[1], label=label, color=color, linewidth=2)
axes[1].set_title('Différentiel de score d\'absence', fontweight='bold')
axes[1].set_xlabel('home_absence - away_absence (+ = dom. plus affaibli)')
axes[1].legend()

plt.suptitle('Impact des Absences de Joueurs Clés sur le Résultat',
             fontsize=13, fontweight='bold')
plt.tight_layout()
plt.show()
```

**💡 Insight :** Les équipes à domicile avec un **score d'absence élevé** (joueurs clés manquants) tendent à perdre plus souvent. Le différentiel d'absence entre les deux équipes est un signal intéressant, bien que modéré — l'absence d'un joueur clé est compensable collectivement, surtout à domicile avec le soutien du public.

## AMÉLIORATION 4 — Calibration Probabiliste

### Contexte

La **calibration** est une dimension souvent négligée mais cruciale : un modèle bien calibré prédit des **probabilités fiables**. Si le modèle prédit 70% de chance de victoire, cela doit se vérifier ~70% du temps.

**Pourquoi c'est important ?**
- Comparaison avec les cotes de paris sportifs
- Prise de décision avec seuil de confiance
- Évaluation via le **Brier Score** (plus bas = mieux calibré)

**Méthode :** `CalibratedClassifierCV` avec régression isotonique (non-paramétrique, plus flexible que Platt scaling).

```python
# ─────────────────────────────────────────────────────────────────────────────
# PRÉPARATION DU DATASET TABULAIRE FINAL
# (nécessaire pour les sections 4, 5, 6)
# ─────────────────────────────────────────────────────────────────────────────

# ── Features classiques du notebook v1 ────────────────────────────────────
matches['pos_diff']         = matches['home_club_position'] - matches['away_club_position']
matches['pos_ratio']        = matches['home_club_position'] / (matches['away_club_position'] + 1)
matches['home_is_top_half'] = (matches['home_club_position'] <= 10).astype(int)
matches['away_is_top_half'] = (matches['away_club_position'] <= 10).astype(int)

# Club features
club_feats = clubs[['club_id','squad_size','average_age',
                    'foreigners_percentage','national_team_players',
                    'stadium_seats','net_transfer_eur']]
matches = matches.merge(
    club_feats.rename(columns={c:f'home_{c}' for c in club_feats.columns if c!='club_id'
                               }).rename(columns={'club_id':'home_club_id'}),
    on='home_club_id', how='left'
)
matches = matches.merge(
    club_feats.rename(columns={c:f'away_{c}' for c in club_feats.columns if c!='club_id'
                               }).rename(columns={'club_id':'away_club_id'}),
    on='away_club_id', how='left'
)

# ── Merge des nouvelles features ───────────────────────────────────────────
dataset = matches.copy()
dataset = dataset.merge(rolling_xg_df,  on='game_id', how='left')
dataset = dataset.merge(absence_df[['game_id','home_absence_score','away_absence_score',
                                     'home_nb_absent_key','away_nb_absent_key','absence_score_diff']],
                        on='game_id', how='left')

# ── Feature engineering de base (forme, H2H, saison) ─────────────────────
# [Ces fonctions sont celles du notebook v1 — réutilisées]

def compute_recent_form(matches_df, n_games=5):
    df = matches_df.sort_values('date').copy().reset_index(drop=True)
    rows = []
    for idx, row in df.iterrows():
        gid=row['game_id']; hid=row['home_club_id']; aid=row['away_club_id']; date=row['date']
        def get_last_n(cid):
            mask=(df['date']<date)&((df['home_club_id']==cid)|(df['away_club_id']==cid))
            return df[mask].tail(n_games)
        def stats(prev, cid):
            if len(prev)==0: return dict(pts=0,gf=0,ga=0,wins=0,draws=0)
            pts=wins=draws=gf=ga=0
            for _,m in prev.iterrows():
                ih=(m['home_club_id']==cid)
                gf+=m['home_club_goals'] if ih else m['away_club_goals']
                ga+=m['away_club_goals'] if ih else m['home_club_goals']
                res=m['results']
                if (ih and res==1) or (not ih and res==-1): pts+=3;wins+=1
                elif res==0: pts+=1;draws+=1
            n=len(prev)
            return dict(pts=pts/n,gf=gf/n,ga=ga/n,wins=wins/n,draws=draws/n)
        sh=stats(get_last_n(hid),hid); sa=stats(get_last_n(aid),aid)
        rows.append({'game_id':gid,
            'home_form_pts':sh['pts'],'home_form_gf':sh['gf'],'home_form_ga':sh['ga'],
            'home_form_wins':sh['wins'],'home_form_draws':sh['draws'],'home_form_n':len(get_last_n(hid)),
            'away_form_pts':sa['pts'],'away_form_gf':sa['gf'],'away_form_ga':sa['ga'],
            'away_form_wins':sa['wins'],'away_form_draws':sa['draws'],'away_form_n':len(get_last_n(aid)),
            'form_pts_diff':sh['pts']-sa['pts'],'form_gf_diff':sh['gf']-sa['gf'],'form_ga_diff':sh['ga']-sa['ga']})
    return pd.DataFrame(rows)

print('⏳ Calcul de la forme récente...')
form_df = compute_recent_form(matches)
dataset = dataset.merge(form_df, on='game_id', how='left')
print(f'✅ Dataset assemblé : {dataset.shape}')
```

```python
# ── Sélection des features & split ────────────────────────────────────────

EXCLUDE = [
    'game_id','season','round','date','results',
    'home_club_goals','away_club_goals',
    'home_club_name','away_club_name',
    'home_club_manager_name','away_club_manager_name',
    'stadium','referee','aggregate','competition_type',
    'home_club_formation','away_club_formation',
    'result_label','total_goals','Unnamed: 0',
    'home_nb_goals','away_nb_goals',
    'xg_diff','penalty_diff','late_goal_diff','discipline_diff',  # <- calculés sur le match en cours
    'home_nb_penalties','away_nb_penalties','home_nb_headers','away_nb_headers',
    'home_nb_late_goals','away_nb_late_goals','home_nb_cards_total','away_nb_cards_total',
    'home_total_approx_xg','away_total_approx_xg',
    'home_nb_freekicks','away_nb_freekicks','home_nb_counter','away_nb_counter',
    'home_nb_longdist','away_nb_longdist','home_nb_own_goals','away_nb_own_goals',
    'home_nb_early_goals','away_nb_early_goals',
    'home_avg_goal_minute','away_avg_goal_minute',
    'home_nb_red_cards','away_nb_red_cards','home_nb_yellow_cards','away_nb_yellow_cards',
]

FEATURE_COLS = [c for c in dataset.columns if c not in EXCLUDE]
print(f'Features sélectionnées : {len(FEATURE_COLS)}')

# Encodage club_id
le = LabelEncoder()
all_clubs = pd.concat([dataset['home_club_id'], dataset['away_club_id'],
                       m2025['home_club_id'], m2025['away_club_id']]).unique()
le.fit(all_clubs)
dataset['home_club_id'] = le.transform(dataset['home_club_id'])
dataset['away_club_id'] = le.transform(dataset['away_club_id'])

X = dataset[FEATURE_COLS].copy()
y = dataset['results'].copy()

imputer = SimpleImputer(strategy='median')
X_imp = pd.DataFrame(imputer.fit_transform(X), columns=X.columns)

train_mask = dataset['season'].values < 2024
test_mask  = dataset['season'].values >= 2024
X_train, X_test = X_imp[train_mask], X_imp[test_mask]
y_train, y_test = y.values[train_mask], y.values[test_mask]

scaler = StandardScaler()
X_train_sc = scaler.fit_transform(X_train)
X_test_sc  = scaler.transform(X_test)

print(f'Train : {X_train.shape[0]} | Test : {X_test.shape[0]}')
```

```python
# ─────────────────────────────────────────────────────────────────────────────
# AMÉLIORATION 4 : CALIBRATION PROBABILISTE
# ─────────────────────────────────────────────────────────────────────────────

# Entraînement du modèle de base (LightGBM)
lgbm_base = lgb.LGBMClassifier(
    n_estimators=300, learning_rate=0.05, max_depth=5,
    num_leaves=31, subsample=0.8, random_state=SEED, n_jobs=-1, verbose=-1
)
lgbm_base.fit(X_train, y_train)

# Modèle calibré (isotonic = régression isotonique non-paramétrique)
lgbm_calibrated = CalibratedClassifierCV(
    lgb.LGBMClassifier(
        n_estimators=300, learning_rate=0.05, max_depth=5,
        num_leaves=31, subsample=0.8, random_state=SEED, n_jobs=-1, verbose=-1
    ),
    method='isotonic', cv=5
)
lgbm_calibrated.fit(X_train, y_train)

# Prédictions et probabilités
proba_base  = lgbm_base.predict_proba(X_test)
proba_calib = lgbm_calibrated.predict_proba(X_test)
pred_base   = lgbm_base.predict(X_test)
pred_calib  = lgbm_calibrated.predict(X_test)

acc_base  = accuracy_score(y_test, pred_base)
acc_calib = accuracy_score(y_test, pred_calib)

# Brier score (somme des carrés des erreurs de probabilité)
label_map = {-1:0, 0:1, 1:2}
y_test_enc = np.array([label_map[y] for y in y_test])
y_test_oh  = np.eye(3)[y_test_enc]  # One-hot

brier_base  = np.mean(np.sum((proba_base  - y_test_oh)**2, axis=1))
brier_calib = np.mean(np.sum((proba_calib - y_test_oh)**2, axis=1))
ll_base     = log_loss(y_test, proba_base)
ll_calib    = log_loss(y_test, proba_calib)

print('📊 Comparaison Calibration :')
print(f'   {'Métrique':30s} {'Non calibré':>15s} {'Calibré':>15s}')
print(f"   {'Accuracy':30s} {acc_base:>15.4f} {acc_calib:>15.4f}")
print(f"   {'Brier Score (↓ mieux)':30s} {brier_base:>15.4f} {brier_calib:>15.4f}")
print(f"   {'Log-Loss (↓ mieux)':30s} {ll_base:>15.4f} {ll_calib:>15.4f}")
```

```python
# ── Courbe de calibration ─────────────────────────────────────────────────

from sklearn.calibration import calibration_curve

fig, axes = plt.subplots(1, 3, figsize=(16, 5))
class_labels = {0: 'Ext. gagne (-1)', 1: 'Nul (0)', 2: 'Dom. gagne (1)'}

for cls_idx, ax in enumerate(axes):
    y_bin = (y_test_enc == cls_idx).astype(int)
    
    # Courbe calibration avant
    frac_pos_b, mean_pred_b = calibration_curve(
        y_bin, proba_base[:, cls_idx], n_bins=8, strategy='quantile'
    )
    # Courbe calibration après
    frac_pos_c, mean_pred_c = calibration_curve(
        y_bin, proba_calib[:, cls_idx], n_bins=8, strategy='quantile'
    )
    
    ax.plot([0,1],[0,1], 'k--', label='Calibration parfaite', linewidth=1.5)
    ax.plot(mean_pred_b, frac_pos_b, 'o-', color='#e74c3c',
            label=f'Non calibré', linewidth=2)
    ax.plot(mean_pred_c, frac_pos_c, 's-', color='#2ecc71',
            label=f'Calibré (isotonic)', linewidth=2)
    ax.set_title(class_labels[cls_idx], fontweight='bold')
    ax.set_xlabel('Probabilité prédite')
    ax.set_ylabel('Fréquence réelle')
    ax.legend(fontsize=8)

plt.suptitle('Courbes de Calibration par Classe\n'
             '(plus proche de la diagonale = mieux calibré)',
             fontsize=13, fontweight='bold')
plt.tight_layout()
plt.show()
```

**💡 Analyse calibration :** Un modèle bien calibré est essentiel pour interpréter les probabilités. Les courbes montrent la différence entre les probabilités prédites et les fréquences réelles. La **régression isotonique** rapproche les courbes de la diagonale idéale, rendant les probabilités plus fiables — particulièrement utile pour la classe "Nul" qui est naturellement difficile à calibrer.

## AMÉLIORATION 5 — Ensemble Learning (Soft Voting)

### Contexte

L'**ensemble learning** combine plusieurs modèles pour réduire la variance et les biais individuels. Le **soft voting** moyenne les **probabilités** (et non les votes discrets) de chaque modèle, tirant profit de leurs forces complémentaires :

| Modèle | Force | Faiblesse |
|--------|-------|----------|
| Logistic Regression | Bien calibré, stable | Linéaire, peu flexible |
| Random Forest | Robuste au bruit | Parfois sur-confiant |
| Gradient Boosting | Séquences de correction | Lent, sensible aux outliers |
| LightGBM | Rapide, précis | Peut over-fitter |

Le vote pondéré permet d'attribuer plus de poids aux modèles les plus performants.

```python
# ─────────────────────────────────────────────────────────────────────────────
# AMÉLIORATION 5 : ENSEMBLE LEARNING — SOFT VOTING PONDÉRÉ
# ─────────────────────────────────────────────────────────────────────────────

# ── Entraînement de chaque modèle individuel ──────────────────────────────
print('⏳ Entraînement des modèles individuels...')

models_dict = {
    'lr'  : LogisticRegression(max_iter=1000, C=0.5, random_state=SEED,
                               multi_class='multinomial'),
    'rf'  : RandomForestClassifier(n_estimators=300, max_depth=8,
                                   min_samples_leaf=10, random_state=SEED, n_jobs=-1),
    'gb'  : GradientBoostingClassifier(n_estimators=200, learning_rate=0.05,
                                        max_depth=4, random_state=SEED),
    'lgbm': lgb.LGBMClassifier(n_estimators=300, learning_rate=0.05, max_depth=5,
                                num_leaves=31, subsample=0.8, random_state=SEED,
                                n_jobs=-1, verbose=-1),
}

individual_accs = {}
for name, model in models_dict.items():
    Xtr = X_train_sc if name == 'lr' else X_train
    Xte = X_test_sc  if name == 'lr' else X_test
    model.fit(Xtr, y_train)
    acc = accuracy_score(y_test, model.predict(Xte))
    individual_accs[name] = acc
    print(f'  {name:6s} → {acc:.4f}')

# ── Soft Voting — 3 variantes ──────────────────────────────────────────────
print('\n⏳ Calcul des ensembles...')

# Récupérer les probabilités de chaque modèle
probas = {}
for name, model in models_dict.items():
    Xte = X_test_sc if name == 'lr' else X_test
    probas[name] = model.predict_proba(Xte)

classes_order = models_dict['lgbm'].classes_  # [-1, 0, 1]

# Variante A : Vote équipondéré (poids égaux)
proba_equal = np.mean([probas[n] for n in probas], axis=0)
pred_equal  = classes_order[np.argmax(proba_equal, axis=1)]
acc_equal   = accuracy_score(y_test, pred_equal)

# Variante B : Vote pondéré par accuracy individuelle
weights_acc = np.array([individual_accs[n] for n in probas])
weights_acc = weights_acc / weights_acc.sum()
proba_weighted = np.average([probas[n] for n in probas], axis=0, weights=weights_acc)
pred_weighted  = classes_order[np.argmax(proba_weighted, axis=1)]
acc_weighted   = accuracy_score(y_test, pred_weighted)

# Variante C : Stacking — meta-modèle LR sur les probabilités
# (utilise un CV sur le train set pour créer des out-of-fold predictions)
from sklearn.model_selection import cross_val_predict

# Créer les meta-features (probas OOF)
meta_features_train = []
meta_features_test  = []
for name, model in models_dict.items():
    Xtr = X_train_sc if name == 'lr' else X_train
    Xte = X_test_sc  if name == 'lr' else X_test
    oof = cross_val_predict(model, Xtr, y_train, cv=5, method='predict_proba',
                            n_jobs=-1)
    model.fit(Xtr, y_train)  # Réentraîner sur tout le train
    test_proba = model.predict_proba(Xte)
    meta_features_train.append(oof)
    meta_features_test.append(test_proba)

X_meta_train = np.hstack(meta_features_train)
X_meta_test  = np.hstack(meta_features_test)

meta_lr = LogisticRegression(C=1.0, max_iter=500, random_state=SEED)
meta_lr.fit(X_meta_train, y_train)
pred_stacking  = meta_lr.predict(X_meta_test)
acc_stacking   = accuracy_score(y_test, pred_stacking)

print(f'\n📊 Résultats Ensemble :')
print(f'  Vote équipondéré     : {acc_equal:.4f}')
print(f'  Vote pondéré (acc)   : {acc_weighted:.4f}')
print(f'  Stacking (meta-LR)   : {acc_stacking:.4f}')
```

```python
# ── Visualisation des probabilités de l'ensemble ──────────────────────────

fig, axes = plt.subplots(1, 3, figsize=(16, 5))

model_names = list(probas.keys()) + ['Ensemble\n(pondéré)', 'Stacking']
accs_all = list(individual_accs.values()) + [acc_weighted, acc_stacking]
colors_bar = ['#95a5a6']*4 + ['#3498db', '#e74c3c']

# Graphique 1 : Comparaison des accuracies
bars = axes[0].bar(model_names, accs_all, color=colors_bar, edgecolor='white')
baseline = (y_test == 1).mean()
axes[0].axhline(baseline, color='black', linestyle='--', linewidth=1.5,
                label=f'Baseline ({baseline:.3f})')
for bar, acc in zip(bars, accs_all):
    axes[0].text(bar.get_x()+bar.get_width()/2, acc+0.002, f'{acc:.4f}',
                 ha='center', fontsize=8, fontweight='bold')
axes[0].set_title('Accuracy : Individuels vs Ensembles', fontweight='bold')
axes[0].set_ylim(0.3, 0.65)
axes[0].legend(fontsize=9)
axes[0].tick_params(axis='x', rotation=20)

# Graphique 2 : Distribution des probabilités (ensemble vs best single)
best_single = max(individual_accs, key=individual_accs.get)
max_proba_best  = np.max(probas[best_single], axis=1)
max_proba_ens   = np.max(proba_weighted, axis=1)

axes[1].hist(max_proba_best, bins=30, alpha=0.6, color='#95a5a6',
             label=f'{best_single} (best single)', edgecolor='white')
axes[1].hist(max_proba_ens,  bins=30, alpha=0.6, color='#3498db',
             label='Ensemble pondéré', edgecolor='white')
axes[1].set_title('Distribution de la confiance maximale\n(proba max prédite)', fontweight='bold')
axes[1].set_xlabel('Probabilité max')
axes[1].legend(fontsize=9)

# Graphique 3 : Poids dans l'ensemble
axes[2].pie(weights_acc,
            labels=[f'{n}\n({w:.2%})' for n, w in zip(model_names, weights_acc)],
            colors=['#e74c3c','#f39c12','#3498db','#2ecc71'],
            autopct='%1.1f%%', startangle=90,
            wedgeprops=dict(edgecolor='white', linewidth=2))
axes[2].set_title('Poids par modèle dans l\'ensemble', fontweight='bold')

plt.suptitle('Ensemble Learning — Analyse Comparative', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

Le vote pondéré par accuracy est une alternative plus simple mais souvent aussi efficace. L'ensemble réduit la variance des prédictions — les erreurs d'un modèle sont compensées par les autres.

## AMÉLIORATION 6 — H2H Pondéré (Décroissance Exponentielle)

### Contexte

Dans la version initiale, tous les matchs H2H historiques étaient traités **de façon égale**. Or, un match joué il y a 8 ans est bien moins informatif qu'un match joué la saison dernière : les effectifs, les entraîneurs et les styles de jeu changent.

**Solution :** Appliquer une **décroissance exponentielle** aux poids des matchs passés :

$$w_i = \lambda^{d_i/365}$$

où $d_i$ est le nombre de jours depuis le match $i$ et $\lambda \in ]0, 1[$ est le facteur de décroissance (typiquement 0.85 pour une demi-vie d'environ 1.5 saison).

```python
# ─────────────────────────────────────────────────────────────────────────────
# AMÉLIORATION 6 : H2H PONDÉRÉ PAR DÉCROISSANCE EXPONENTIELLE
# ─────────────────────────────────────────────────────────────────────────────

DECAY_FACTOR = 0.85   # λ : poids réduit de 15% par an

def compute_h2h_weighted(matches_df, n=10, decay=DECAY_FACTOR):
    """
    H2H avec pondération temporelle exponentielle.
    Les matchs récents ont un poids plus élevé.
    """
    df = matches_df.sort_values('date').copy()
    rows = []
    
    for _, row in df.iterrows():
        hid  = row['home_club_id']
        aid  = row['away_club_id']
        date = row['date']
        
        prev = df[
            (df['date'] < date) & (
                ((df['home_club_id']==hid) & (df['away_club_id']==aid)) |
                ((df['home_club_id']==aid) & (df['away_club_id']==hid))
            )
        ].tail(n)
        
        if len(prev) == 0:
            rows.append({
                'game_id':row['game_id'],
                'h2h_w_home_wins':0.33, 'h2h_w_away_wins':0.30, 'h2h_w_draws':0.26,
                'h2h_w_home_gf':1.4, 'h2h_w_away_gf':1.1, 'h2h_n':0,
                'h2h_w_home_xg':0.0, 'h2h_recency':0.0,
                # Comparaison : H2H non pondéré
                'h2h_home_wins':0.33, 'h2h_away_wins':0.30, 'h2h_draws':0.26,
            })
            continue
        
        # Calcul des poids temporels
        days_ago = [(date - m_date).days for m_date in prev['date']]
        weights  = np.array([decay ** (d / 365.0) for d in days_ago])
        weights  = weights / weights.sum()  # Normalisation
        
        # Métriques pondérées
        hw_w = aw_w = d_w = hgf_w = agf_w = hxg_w = 0.0
        # Métriques non pondérées (pour comparaison)
        hw_u = aw_u = d_u = 0.0
        
        for (_, m), w in zip(prev.iterrows(), weights):
            is_h = (m['home_club_id'] == hid)
            gf = m['home_club_goals'] if is_h else m['away_club_goals']
            ga = m['away_club_goals'] if is_h else m['home_club_goals']
            res = m['results']
            
            home_won = (is_h and res==1) or (not is_h and res==-1)
            away_won = (is_h and res==-1) or (not is_h and res==1)
            is_draw  = (res == 0)
            
            hgf_w += gf * w; agf_w += ga * w
            if home_won: hw_w += w; hw_u += 1
            elif away_won: aw_w += w; aw_u += 1
            else: d_w += w; d_u += 1
            
            # xG approximatif de l'équipe dom dans ce H2H
            xg_col = 'home_total_approx_xg' if is_h else 'away_total_approx_xg'
            xg_val = m.get(xg_col, 0) or 0
            hxg_w += xg_val * w
        
        n_prev = len(prev)
        rows.append({
            'game_id'          : row['game_id'],
            'h2h_w_home_wins'  : hw_w,
            'h2h_w_away_wins'  : aw_w,
            'h2h_w_draws'      : d_w,
            'h2h_w_home_gf'    : hgf_w,
            'h2h_w_away_gf'    : agf_w,
            'h2h_w_home_xg'    : hxg_w,
            'h2h_n'            : n_prev,
            'h2h_recency'      : weights[-1] if len(weights) > 0 else 0,  # Poids du match le + récent
            # Non pondéré
            'h2h_home_wins'    : hw_u / n_prev,
            'h2h_away_wins'    : aw_u / n_prev,
            'h2h_draws'        : d_u  / n_prev,
        })
    
    return pd.DataFrame(rows)

print(f'⏳ Calcul H2H pondéré (λ={DECAY_FACTOR})...')
h2h_weighted_df = compute_h2h_weighted(matches)
print(f'✅ H2H pondéré calculé : {h2h_weighted_df.shape}')
display(h2h_weighted_df[h2h_weighted_df['h2h_n']>0].head(5))
```

```python
# ── Visualisation : Impact de la pondération temporelle ──────────────────

fig, axes = plt.subplots(1, 3, figsize=(16, 5))

# Graphique 1 : Poids selon l'ancienneté
days = np.arange(0, 3651, 30)
for lam, color, label in [
    (0.70, '#e74c3c', 'λ=0.70 (décroissance rapide, ½vie~10 mois)'),
    (0.85, '#3498db', 'λ=0.85 (décroissance modérée, ½vie~18 mois)'),
    (0.95, '#2ecc71', 'λ=0.95 (décroissance lente, ½vie~4 ans)'),
]:
    w = lam ** (days / 365)
    axes[0].plot(days / 365, w, color=color, linewidth=2, label=label)
axes[0].set_title('Schémas de décroissance exponentielle', fontweight='bold')
axes[0].set_xlabel('Années depuis le match')
axes[0].set_ylabel('Poids relatif')
axes[0].legend(fontsize=8)
axes[0].axhline(0.5, linestyle='--', color='gray', alpha=0.5, label='Poids 0.5')

# Graphique 2 : H2H pondéré vs non pondéré pour les victoires dom
h2h_with_res = h2h_weighted_df.merge(matches[['game_id','results']], on='game_id')
h2h_with_res = h2h_with_res[h2h_with_res['h2h_n'] > 0]

for res, label, color in [(1,'Dom. gagne','#2ecc71'),(0,'Nul','#f39c12'),(-1,'Ext. gagne','#e74c3c')]:
    sub = h2h_with_res[h2h_with_res['results']==res]
    sub['h2h_w_home_wins'].plot(kind='kde', ax=axes[1], label=label, color=color, linewidth=2)
axes[1].set_title('H2H wins pondéré — Dom. selon résultat', fontweight='bold')
axes[1].set_xlabel('h2h_w_home_wins')
axes[1].legend()

# Graphique 3 : Corrélation pondéré vs non pondéré
h2h_both = h2h_with_res[['h2h_home_wins','h2h_w_home_wins']].dropna()
axes[2].scatter(h2h_both['h2h_home_wins'], h2h_both['h2h_w_home_wins'],
                alpha=0.3, color='#9b59b6', s=10)
axes[2].plot([0,1],[0,1],'k--', linewidth=1.5, label='Identique')
axes[2].set_title('H2H pondéré vs non pondéré', fontweight='bold')
axes[2].set_xlabel('H2H wins (non pondéré)')
axes[2].set_ylabel('H2H wins (pondéré)')
axes[2].legend()

plt.suptitle('H2H Pondéré — Analyse de l\'Impact Temporel',
             fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

**💡 Analyse H2H pondéré :** La dispersion entre H2H pondéré et non pondéré révèle des cas où la pondération **change significativement l'interprétation** : une équipe qui dominait une autre il y a 5 ans mais a perdu les dernières confrontations voit son score H2H pondéré diminuer, reflétant mieux la réalité sportive actuelle.

```python
# ─────────────────────────────────────────────────────────────────────────────
# DATASET FINAL avec TOUTES les améliorations
# ─────────────────────────────────────────────────────────────────────────────

dataset_v2 = dataset.copy()
dataset_v2 = dataset_v2.merge(
    h2h_weighted_df[['game_id','h2h_w_home_wins','h2h_w_away_wins',
                     'h2h_w_draws','h2h_w_home_gf','h2h_w_away_gf',
                     'h2h_w_home_xg','h2h_recency','h2h_n',
                     'h2h_home_wins','h2h_away_wins','h2h_draws']],
    on='game_id', how='left'
)

FEATURE_COLS_V2 = [c for c in dataset_v2.columns if c not in (
    EXCLUDE + ['home_total_approx_xg','away_total_approx_xg',
               'home_nb_penalties','away_nb_penalties','home_nb_headers',
               'away_nb_headers','home_nb_late_goals','away_nb_late_goals',
               'home_nb_cards_total','away_nb_cards_total',
               'home_nb_freekicks','away_nb_freekicks',
               'home_nb_counter','away_nb_counter',
               'home_nb_longdist','away_nb_longdist',
               'home_nb_own_goals','away_nb_own_goals',
               'home_nb_early_goals','away_nb_early_goals',
               'home_avg_goal_minute','away_avg_goal_minute',
               'home_nb_red_cards','away_nb_red_cards',
               'home_nb_yellow_cards','away_nb_yellow_cards',
               'xg_diff','penalty_diff','late_goal_diff','discipline_diff']
)]

X_v2 = dataset_v2[FEATURE_COLS_V2].copy()
y_v2 = dataset_v2['results'].copy()

imputer_v2 = SimpleImputer(strategy='median')
X_v2_imp = pd.DataFrame(imputer_v2.fit_transform(X_v2), columns=X_v2.columns)

train_mask_v2 = dataset_v2['season'].values < 2024
test_mask_v2  = dataset_v2['season'].values >= 2024

Xtr_v2 = X_v2_imp[train_mask_v2]; Xte_v2 = X_v2_imp[test_mask_v2]
ytr_v2 = y_v2.values[train_mask_v2]; yte_v2 = y_v2.values[test_mask_v2]

print(f'Dataset v2 : {X_v2_imp.shape[1]} features | Train {Xtr_v2.shape[0]} | Test {Xte_v2.shape[0]}')
```

```python
# ─────────────────────────────────────────────────────────────────────────────
# COMPARAISON FINALE : tous les modèles
# ─────────────────────────────────────────────────────────────────────────────

baseline = (yte_v2 == 1).mean()
results_final = {}

def test_model(name, model, Xtr, ytr, Xte, yte):
    model.fit(Xtr, ytr)
    preds = model.predict(Xte)
    probas = model.predict_proba(Xte) if hasattr(model, 'predict_proba') else None
    acc = accuracy_score(yte, preds)
    results_final[name] = {'acc': acc, 'preds': preds, 'probas': probas, 'model': model}
    print(f'  {name:40s} → {acc:.4f}')

print('📊 Évaluation finale (dataset avec toutes les améliorations):\n')

# Modèles de base
test_model('Logistic Regression',
    LogisticRegression(max_iter=1000, C=0.5, random_state=SEED, multi_class='multinomial'),
    scaler.fit_transform(Xtr_v2), ytr_v2, scaler.transform(Xte_v2), yte_v2)

test_model('Random Forest',
    RandomForestClassifier(n_estimators=300, max_depth=8, min_samples_leaf=10,
                           random_state=SEED, n_jobs=-1),
    Xtr_v2, ytr_v2, Xte_v2, yte_v2)

test_model('Gradient Boosting',
    GradientBoostingClassifier(n_estimators=200, learning_rate=0.05,
                                max_depth=4, random_state=SEED),
    Xtr_v2, ytr_v2, Xte_v2, yte_v2)

test_model('LightGBM',
    lgb.LGBMClassifier(n_estimators=300, learning_rate=0.05, max_depth=5,
                       num_leaves=31, subsample=0.8, random_state=SEED,
                       n_jobs=-1, verbose=-1),
    Xtr_v2, ytr_v2, Xte_v2, yte_v2)

test_model('LightGBM Calibré',
    CalibratedClassifierCV(
        lgb.LGBMClassifier(n_estimators=300, learning_rate=0.05, max_depth=5,
                           num_leaves=31, subsample=0.8, random_state=SEED,
                           n_jobs=-1, verbose=-1),
        method='isotonic', cv=5
    ),
    Xtr_v2, ytr_v2, Xte_v2, yte_v2)

# Ensemble
probas_ens = [results_final[n]['probas'] for n in
              ['Logistic Regression','Random Forest','Gradient Boosting','LightGBM']
              if results_final[n]['probas'] is not None]
accs_ens   = [results_final[n]['acc'] for n in
              ['Logistic Regression','Random Forest','Gradient Boosting','LightGBM']]
w_ens = np.array(accs_ens); w_ens /= w_ens.sum()
proba_ens_final = np.average(probas_ens, axis=0, weights=w_ens)
# classes order dépend du modèle, on récupère depuis RF
cls_ord = results_final['Random Forest']['model'].classes_
pred_ens_final = cls_ord[np.argmax(proba_ens_final, axis=1)]
acc_ens_final  = accuracy_score(yte_v2, pred_ens_final)
results_final['Ensemble Pondéré'] = {'acc': acc_ens_final, 'preds': pred_ens_final}
print(f'  {"Ensemble Pondéré":40s} → {acc_ens_final:.4f}')

# LSTM (déjà calculé)
print(f'  {"LSTM Double-Branche":40s} → {acc_lstm:.4f}')
results_final['LSTM Double-Branche'] = {'acc': acc_lstm, 'preds': lstm_preds_orig}

print(f'\n  {"Baseline (toujours dom.)":40s} → {baseline:.4f}')
```

## Génération des Prédictions Finales

```python
# ─────────────────────────────────────────────────────────────────────────────
# PRÉDICTIONS 2025 — Meilleur modèle entraîné sur 100% des données
# ─────────────────────────────────────────────────────────────────────────────

# On réentraîne le meilleur modèle sur toutes les données
final_model_v2 = lgb.LGBMClassifier(
    n_estimators=300, learning_rate=0.05, max_depth=5,
    num_leaves=31, subsample=0.8, random_state=SEED, n_jobs=-1, verbose=-1
)
final_model_v2.fit(X_v2_imp, y_v2)

# Construction des features 2025
# (Réutilise la logique du notebook v1 + nouvelles features)
m2025_pred = m2025.copy()
m2025_pred['season'] = 2024
m2025_pred['date'] = pd.to_datetime(m2025_pred['date'])

# Position de classement (dernière connue saison 2024)
last_pos = matches[matches['season']==2024][['home_club_id','home_club_position']]\
    .rename(columns={'home_club_id':'club_id','home_club_position':'position'})\
    .groupby('club_id')['position'].last().reset_index()

m2025_pred = m2025_pred.merge(
    last_pos.rename(columns={'club_id':'home_club_id','position':'home_club_position'}),
    on='home_club_id', how='left'
)
m2025_pred = m2025_pred.merge(
    last_pos.rename(columns={'club_id':'away_club_id','position':'away_club_position'}),
    on='away_club_id', how='left'
)
m2025_pred['home_club_position'] = m2025_pred['home_club_position'].fillna(10)
m2025_pred['away_club_position'] = m2025_pred['away_club_position'].fillna(10)
m2025_pred['pos_diff']           = m2025_pred['home_club_position'] - m2025_pred['away_club_position']
m2025_pred['pos_ratio']          = m2025_pred['home_club_position'] / (m2025_pred['away_club_position'] + 1)
m2025_pred['home_is_top_half']   = (m2025_pred['home_club_position'] <= 10).astype(int)
m2025_pred['away_is_top_half']   = (m2025_pred['away_club_position'] <= 10).astype(int)

# Valuation, club features, forme, H2H pondéré...
m2025_pred = add_valuation_features(m2025_pred, club_val_season)

club_feats_pred = clubs[['club_id','squad_size','average_age',
                          'foreigners_percentage','national_team_players',
                          'stadium_seats','net_transfer_eur']]
m2025_pred = m2025_pred.merge(
    club_feats_pred.rename(columns={c:f'home_{c}' for c in club_feats_pred.columns if c!='club_id'
                                    }).rename(columns={'club_id':'home_club_id'}),
    on='home_club_id', how='left'
)
m2025_pred = m2025_pred.merge(
    club_feats_pred.rename(columns={c:f'away_{c}' for c in club_feats_pred.columns if c!='club_id'
                                    }).rename(columns={'club_id':'away_club_id'}),
    on='away_club_id', how='left'
)

# Encodage
m2025_pred['home_club_id'] = le.transform(m2025_pred['home_club_id'])
m2025_pred['away_club_id'] = le.transform(m2025_pred['away_club_id'])

# Sélection des features et imputation
missing_cols = [c for c in FEATURE_COLS_V2 if c not in m2025_pred.columns]
for c in missing_cols:
    m2025_pred[c] = 0

X_2025_v2 = m2025_pred[FEATURE_COLS_V2].copy()
X_2025_v2_imp = pd.DataFrame(imputer_v2.transform(X_2025_v2), columns=X_2025_v2.columns)

# Prédictions
preds_2025 = final_model_v2.predict(X_2025_v2_imp)
probas_2025 = final_model_v2.predict_proba(X_2025_v2_imp)

output = pd.DataFrame({
    'game_id': m2025['game_id'],
    'results': preds_2025
})
output.to_csv('predictions_2025_v2.csv', index=False)

print('✅ Fichier predictions_2025_v2.csv généré !')
print(f'Distribution des prédictions :\n{pd.Series(preds_2025).value_counts()}')
display(output.head(10))
```

```python
# ── Visualisation finale des prédictions ──────────────────────────────────

fig, axes = plt.subplots(1, 3, figsize=(16, 5))

label_plot = {1:'Dom. gagne', 0:'Nul', -1:'Ext. gagne'}
pred_counts = pd.Series(preds_2025).map(label_plot).value_counts()
colors = ['#2ecc71','#f39c12','#e74c3c']

# Distribution
axes[0].bar(pred_counts.index, pred_counts.values, color=colors, edgecolor='white')
for i,(l,v) in enumerate(pred_counts.items()):
    axes[0].text(i, v+1, f'{v}\n({v/len(preds_2025)*100:.1f}%)',
                 ha='center', fontweight='bold')
axes[0].set_title('Distribution des prédictions 2025-2026', fontweight='bold')

# Comparaison historique vs prédictions
hist_d = matches['results'].map(label_plot).value_counts(normalize=True)*100
pred_d = pd.Series(preds_2025).map(label_plot).value_counts(normalize=True)*100
x = np.arange(3); w = 0.35
axes[1].bar(x-w/2, [hist_d.get(l,0) for l in pred_d.index], w,
            label='Historique', color='#3498db', alpha=0.8)
axes[1].bar(x+w/2, pred_d.values, w,
            label='Prédictions', color='#9b59b6', alpha=0.8)
axes[1].set_xticks(x)
axes[1].set_xticklabels(pred_d.index)
axes[1].set_title('Historique vs Prédictions', fontweight='bold')
axes[1].legend()

# Distribution des probabilités maximales (confiance)
max_probas = np.max(probas_2025, axis=1)
axes[2].hist(max_probas, bins=25, color='#3498db', edgecolor='white', alpha=0.8)
axes[2].axvline(max_probas.mean(), color='#e74c3c', linestyle='--',
                label=f'Moyenne : {max_probas.mean():.3f}')
axes[2].set_title('Confiance du modèle (proba max)', fontweight='bold')
axes[2].set_xlabel('Probabilité maximale prédite')
axes[2].legend()

plt.suptitle('Prédictions Finales — Ligue 1 Saison 2025-2026 (Version Améliorée)',
             fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

## Conclusion

### Bilan des améliorations

| Amélioration | Impact attendu | Implémentation |
|---|---|---|
| **Proxies xG** | Capture la qualité offensive au-delà des buts | ✅ Types de tirs, pénaltys, têtes, timing |
| **LSTM** | Patterns temporels, momentum | ✅ Architecture double-branche |
| **Absences joueurs clés** | Impact des indisponibilités | ✅ Score d'impact × présence lineup |
| **Calibration probabiliste** | Probabilités fiables, Brier score | ✅ Isotonic regression (5-fold) |
| **Ensemble pondéré + Stacking** | Réduction variance, complémentarité | ✅ Vote pondéré + meta-LR |
| **H2H pondéré (λ=0.85)** | Matchs récents plus informatifs | ✅ Décroissance exponentielle |

### Limites persistantes

- Le football reste **fondamentalement imprévisible** : même les modèles les plus sophistiqués plafonnent autour de 55-60% d'accuracy
- L'absence de données xG réelles (Opta/Statsbomb) limite la précision des proxies
- Le LSTM souffre du **petit nombre de séquences** disponibles par club
- Les **compositions 2025** ne sont pas connues à l'avance, ce qui limite l'utilisation des features de lineup

### Meilleur modèle final

Le meilleur modèle est l'**Ensemble Pondéré** (ou LightGBM Calibré), combinant :
1. Features de forme récente (5 matchs)
2. H2H pondéré temporellement
3. Proxies xG issus des events
4. Scores d'absence joueurs clés
5. Valeur marchande des effectifs
6. Position au classement et statistiques de saison
